```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
tea_data = pd.read_csv("aps_tea_south_africa_cleaned.xls")
```

```python
tea_data.head()
```

```python
tea_data.info()
```

```python
tea_data.isnull().sum()
```

```python
# Remove missing TEA values
tea_values = tea_data["TEA"].dropna()

# Create histogram
plt.figure(figsize=(10, 6))

plt.hist(
    tea_values,
    bins=6,
    edgecolor="black"
)

plt.title("Distribution of Total Early-Stage Entrepreneurial Activity (TEA %)")
plt.xlabel("TEA (%)")
plt.ylabel("Frequency")

plt.grid(axis="y", alpha=0.3)

plt.show()
```

```python
diet_data = pd.read_csv("coahd_cleaned.xls")
```

```python
diet_data.head()
```

```python
diet_data.columns
```

```python
cost_data = diet_data[
    diet_data["INDICATOR_LABEL"].str.startswith("Cost of a healthy diet")
].copy()
```

```python
unaffordable_data = diet_data[
    diet_data["INDICATOR_LABEL"].str.startswith(
        "Percentage of the population unable"
    )
].copy()
```

```python
cost_data[["Year", "Value"]]
```

```python
unaffordable_data[["Year", "Value"]]
```

```python
fig, ax1 = plt.subplots(figsize=(12, 6))

# Healthy diet cost
ax1.plot(
    cost_data["Year"],
    cost_data["Value"],
    marker="o",
    linewidth=2,
    label="Healthy Diet Cost"
)

ax1.set_xlabel("Year")
ax1.set_ylabel("Cost of Healthy Diet (PPP dollars per person per day)")

ax1.grid(True, alpha=0.3)

# Create second y-axis
ax2 = ax1.twinx()

# Percentage unable to afford
ax2.plot(
    unaffordable_data["Year"],
    unaffordable_data["Value"],
    marker="s",
    linestyle="--",
    linewidth=2,
    label="Population Unable to Afford"
)

ax2.set_ylabel("Population Unable to Afford Healthy Diet (%)")

plt.title(
    "Trends in Healthy Diet Cost and Unaffordability in South Africa"
)

plt.show()
```

```python
plt.figure(figsize=(12, 6))

bars = plt.bar(
    tea_data["Year"],
    tea_data["TEA"]
)

plt.title(
    "Total Early-Stage Entrepreneurial Activity (TEA) in South Africa"
)

plt.xlabel("Year")
plt.ylabel("TEA (%)")

plt.xticks(tea_data["Year"], rotation=45)

plt.grid(axis="y", alpha=0.3)

# Add values above bars
for bar, value in zip(bars, tea_data["TEA"]):
    if pd.notna(value):
        plt.text(
            bar.get_x() + bar.get_width() / 2,
            value + 0.3,
            f"{value:.2f}%",
            ha="center"
        )

plt.tight_layout()

plt.show()
```
